```python
#18. Write a program to Reverse a number.
num = int(input("Enter number:"))
reverse = 0
while num!=0:
    last = num%10
    reverse = reverse*10+last
    num=num//10
print("Reverse number =", reverse)
```

```python
#19. Write a program to check if the number is palindrome.
num = int(input("Enter number"))
reverse = 0
x=num
while num!=0:
    last = num%10
    reverse = reverse*10+last
    num=num//10
if x==reverse:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```

```python
#22. Write a program to find armstrong number.
num = int(input("Enter number:"))
y=num
sum = 0
count = len(str(num))
while num!=0:
    digit = num%10
    sum+=digit**count
    num//=10
if y==sum:
    print("Armstrong number")
else:
    print("Not a Armstrong number") 
```

```python
#26. Write a program to count vowels, consonants, spaces, numbers and specialSymbols in a string.
str = "Hello World!"
vowels = 0
consonants = 0
spaces = 0
numbers = 0
spcsymbol = 0
for x in str:
    if (x>="a" and x<="z") or (x>="A" and x<="Z"):
        if x in "AEIOUaeiou":
            vowels+=1
        else:
            consonants+=1
    elif (x>="0" and x<="9"):
        numbers+=1
    elif x==" ":
        spaces+=1
    else:
        spcsymbol+=1

print(f"Vowels = {vowels}, Consonants = {consonants}, spaces = {spaces}, Numbers = {numbers}, spcsymbol = {spcsymbol}")
```

```python
#32. Write a program to find the second largest number.
lst = [1,4,2,12,5,8,10]
largest = lst[0]
secondLargest = lst[-1]
for i in range(len(lst)):
    if lst[i]>largest:
        secondLargest = largest
        largest = lst[i]
    elif lst[i]>secondLargest and lst[i]!=largest:
        secondLargest = lst[i]
print(secondLargest)
```
